Este notebook carrega e analiza os dados do programa PEDE (Programa de Desenvolvimento Educacional) dos anos 2022, 2023 e 2024.

```python
# Importar bibliotecas necessárias
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import sys, os

sys.path.append(os.path.abspath('..'))

# importar funções úteis do módulo de preprocessing
from src.preprocessing import *
from src.feature_engineering import *

# Configurar visualização
%matplotlib inline
plt.style.use('seaborn-v0_8-whitegrid')
plt.rcParams['figure.figsize'] = (12, 6)
plt.rcParams['font.size'] = 12
```

```python
# Carregar o arquivo Excel
file_path = r'C:\Users\erick\Documents\FIAP\TC\FASE5_DATATHON\Base_pede.xlsx'

# Carregar os dados de cada ano
df_2022 = pd.read_excel(file_path, sheet_name='PEDE2022')
df_2023 = pd.read_excel(file_path, sheet_name='PEDE2023')
df_2024 = pd.read_excel(file_path, sheet_name='PEDE2024')

print("=" * 60)
print("INFORMAÇÕES DO DATASET")
print("=" * 60)
print(f"\nPEDE2022: {df_2022.shape[0]} linhas, {df_2022.shape[1]} colunas")
print(f"PEDE2023: {df_2023.shape[0]} linhas, {df_2023.shape[1]} colunas")
print(f"PEDE2024: {df_2024.shape[0]} linhas, {df_2024.shape[1]} colunas")
```

```python
# padroniza para 2022
df22, resumo_22 = padronizar_colunas(df_2022, column_mapping, year=2022)

print(resumo_22['renomeadas'])
print(resumo_22['excluidas'])
```

```python
# padroniza para 2023
df23, resumo_23 = padronizar_colunas(df_2023, column_mapping, year=2023)

print(resumo_23['renomeadas'])
print(resumo_23['excluidas'])
```

```python
# padroniza para 2024
df24, resumo_24 = padronizar_colunas(df_2024, column_mapping, year=2024)

print(resumo_24['renomeadas'])
print(resumo_24['excluidas'])
```

```python
#Unificar os datasets padronizados
df_pede = pd.concat([df22, df23, df24], ignore_index=True)
df_pede[df_pede["ANO_BASE"] == 2024].head()
```

```python
# Tipos de dados - PEDE
print("\nTipos de dados - PEDE:")
print(df_pede.info())
```

```python
df_schema = fix_idade(df_pede)
df_schema, schema_errors = enforce_schema(df_pede, schema= EXPECTED_SCHEMA)

NUMERIC_COLS = [
    "INDE", "IAA", "IEG", "IPS", "IDA", "IPP", "IPV", "IAN",
    "DEFASAGEM", "NOTA_PORT", "NOTA_MAT", "NOTA_ING",
    "ANO_INGRESSO", "QTDE_AVAL"
]

CATEGORICAL_COLS = [
    "FASE", "TURMA", "PEDRA", "NIVEL_IDEAL"
]

df_pede, missing_summary = handle_missing_values(
    df_schema,
    numeric_cols=NUMERIC_COLS,
    categorical_cols=CATEGORICAL_COLS,
    add_missing_flags=True
)
```

```python
# Tipos de dados - PEDE
print("\nTipos de dados - PEDE:")
print(df_pede.info())
```

```python
# Estatísticas descritivas - PEDE2024
print("\nEstatísticas descritivas - PEDE:")
df_pede.describe()
```

```python
# Verificar valores nulos - PEDE
print("\nValores nulos por coluna - PEDE:")
null_counts = df_pede.isnull().sum()
null_pct = (null_counts / len(df_pede)) * 100
null_df = pd.DataFrame({'Quantidade': null_counts, 'Porcentagem': null_pct})
print(null_df[null_df['Quantidade'] > 0].sort_values('Quantidade', ascending=False))
```

```python
df_fe = df_pede.copy()

df_fe = bin_idade(df_fe)
df_fe = encode_pedra(df_fe)
df_fe = create_basic_features(df_fe)
```

```python
df_fe.head()
df_fe[[
    "IDADE_FAIXA",
    "PEDRA",
    "PEDRA_ORDINAL",
    "DEFASADO",
    "NOTA_MEDIA",
    "TEMPO_PM",
    "ALTO_INDE"
]].sample(5)
```

```python
inde_col = 'INDE'

# converter para numérico, coercing errors para NaN
inde_series = pd.to_numeric(df_pede.get(inde_col, pd.Series(dtype='float')), errors='coerce')
inde_series = inde_series.dropna()

fig, axes = plt.subplots(1, 2, figsize=(14, 5))

# Histograma
axes[0].hist(inde_series, bins=30, edgecolor='black', alpha=0.7)
axes[0].set_xlabel('INDE')
axes[0].set_ylabel('Frequência')
axes[0].set_title('Distribuição do INDE')

# Boxplot
axes[1].boxplot(inde_series)
axes[1].set_ylabel('INDE')
axes[1].set_title('Boxplot do INDE')

plt.tight_layout()
plt.show()

print(f"\nEstatísticas do INDE:")
print(inde_series.describe())
```

```python
# Análise da variável Pedra
pedra_col = 'PEDRA'

# Contagem de cada categoria
pedra_counts = df_pede[pedra_col].value_counts()

plt.figure(figsize=(10, 6))
pedra_counts.plot(kind='bar', color='steelblue', edgecolor='black')
plt.xlabel('Pedra')
plt.ylabel('Quantidade')
plt.title('Distribuição da Pedra')
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()

print("\nContagem por Pedra:")
print(pedra_counts)
```

```python
# Análise da variável Pedra
idade_col = 'IDADE_FAIXA'

# Contagem de cada categoria
pedra_counts = df_fe[idade_col].value_counts()

plt.figure(figsize=(10, 6))
pedra_counts.plot(kind='bar', color='steelblue', edgecolor='black')
plt.xlabel('Pedra')
plt.ylabel('Quantidade')
plt.title('Distribuição da Pedra')
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()

print("\nContagem por Pedra:")
print(pedra_counts)
```

```python
# Comparação entre anos - Quantidade de registros
years = ['2022', '2023', '2024']
counts = [len(df_pede[df_pede["ANO_BASE"] == 2022]), len(df_pede[df_pede["ANO_BASE"] == 2023]), len(df_pede[df_pede["ANO_BASE"] == 2024])]

plt.figure(figsize=(10, 6))
plt.bar(years, counts, color=['#2ecc71', '#3498db', '#9b59b6'], edgecolor='black')
plt.xlabel('Ano')
plt.ylabel('Quantidade de Registros')
plt.title('Quantidade de Registros por Ano')
for i, v in enumerate(counts):
    plt.text(i, v + 50, str(v), ha='center', fontweight='bold')
plt.tight_layout()
plt.show()
```

```python
# Correlação entre variáveis numéricas
print(f"Colunas numéricas: {list(NUMERIC_COLS)}")

plt.figure(figsize=(14, 10))
sns.heatmap(df_pede[NUMERIC_COLS].corr(), annot=True, cmap='coolwarm', center=0, fmt='.2f')
plt.title('Mapa de Correlação - Variáveis Numéricas')
plt.tight_layout()
plt.show()
```

```python
# Análise por Fase/Turma
fase_counts = df_pede['NIVEL_IDEAL'].value_counts()

plt.figure(figsize=(10, 6))
fase_counts.plot(kind='bar', color='coral', edgecolor='black')
plt.xlabel('Fase')
plt.ylabel('Quantidade')
plt.title('Distribuição por Fase')
plt.xticks(rotation=0)
plt.tight_layout()
plt.show()
```
