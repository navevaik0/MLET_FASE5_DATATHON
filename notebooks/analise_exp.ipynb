# Análise Exploratória - DATATHON 2024

Este notebook carrega e analiza os dados do programa PEDE (Programa de Développemento Educacional) dos anos 2022, 2023 e 2024.

```python
# Importar bibliotecas necessárias
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

# Configurar visualização
%matplotlib inline
plt.style.use('seaborn-v0_8-whitegrid')
plt.rcParams['figure.figsize'] = (12, 6)
plt.rcParams['font.size'] = 12
```

```python
# Carregar o arquivo Excel
file_path = 'BASE DE DADOS PEDE 2024 - DATATHON.xlsx'

# Verificar as abas disponíveis
xl = pd.ExcelFile(file_path)
print("Abas disponíveis no arquivo Excel:")
for sheet in xl.sheet_names:
    print(f"  - {sheet}")
```

```python
# Carregar os dados de cada ano
df_2022 = pd.read_excel(file_path, sheet_name='PEDE2022')
df_2023 = pd.read_excel(file_path, sheet_name='PEDE2023')
df_2024 = pd.read_excel(file_path, sheet_name='PEDE2024')

print("=" * 60)
print("INFORMAÇÕES DO DATASET")
print("=" * 60)
print(f"\nPEDE2022: {df_2022.shape[0]} linhas, {df_2022.shape[1]} colunas")
print(f"PEDE2023: {df_2023.shape[0]} linhas, {df_2023.shape[1]} colunas")
print(f"PEDE2024: {df_2024.shape[0]} linhas, {df_2024.shape[1]} colunas")
```

```python
# Visualizar as primeiras linhas do dataset 2024
print("Primeiras 5 linhas do dataset PEDE2024:")
df_2024.head()
```

```python
# Todas as colunas do dataset 2024
print("Todas as colunas do dataset PEDE2024:")
for i, col in enumerate(df_2024.columns, 1):
    print(f"{i:2d}. {col}")
```

```python
# Tipos de dados - PEDE2024
print("\nTipos de dados - PEDE2024:")
print(df_2024.dtypes)
```

```python
# Estatísticas descritivas - PEDE2024
print("\nEstatísticas descritivas - PEDE2024:")
df_2024.describe()
```

```python
# Verificar valores nulos - PEDE2024
print("\nValores nulos por coluna - PEDE2024:")
null_counts = df_2024.isnull().sum()
null_pct = (null_counts / len(df_2024)) * 100
null_df = pd.DataFrame({'Quantidade': null_counts, 'Porcentagem': null_pct})
print(null_df[null_df['Quantidade'] > 0].sort_values('Quantidade', ascending=False))
```

```python
# Análise da variável INDE 2024 (Índice de Desenvolvimento Educacional)
if 'INDE 2024' in df_2024.columns:
    inde_col = 'INDE 2024'
    
    fig, axes = plt.subplots(1, 2, figsize=(14, 5))
    
    # Histograma
    axes[0].hist(df_2024[inde_col].dropna(), bins=30, edgecolor='black', alpha=0.7)
    axes[0].set_xlabel('INDE 2024')
    axes[0].set_ylabel('Frequência')
    axes[0].set_title('Distribuição do INDE 2024')
    
    # Boxplot
    axes[1].boxplot(df_2024[inde_col].dropna())
    axes[1].set_ylabel('INDE 2024')
    axes[1].set_title('Boxplot do INDE 2024')
    
    plt.tight_layout()
    plt.show()
    
    print(f"\nEstatísticas do INDE 2024:")
    print(df_2024[inde_col].describe())
```

```python
# Análise da variável Pedra 2024
if 'Pedra 2024' in df_2024.columns:
    pedra_col = 'Pedra 2024'
    
    # Contagem de cada categoria
    pedra_counts = df_2024[pedra_col].value_counts()
    
    plt.figure(figsize=(10, 6))
    pedra_counts.plot(kind='bar', color='steelblue', edgecolor='black')
    plt.xlabel('Pedra 2024')
    plt.ylabel('Quantidade')
    plt.title('Distribuição da Pedra 2024')
    plt.xticks(rotation=45)
    plt.tight_layout()
    plt.show()
    
    print("\nContagem por Pedra 2024:")
    print(pedra_counts)
```

```python
# Análise de Gênero
if 'Gênero' in df_2024.columns:
    genero_counts = df_2024['Gênero'].value_counts()
    
    plt.figure(figsize=(8, 6))
    genero_counts.plot(kind='pie', autopct='%1.1f%%', startangle=90)
    plt.ylabel('')
    plt.title('Distribuição por Gênero - 2024')
    plt.tight_layout()
    plt.show()
    
    print("\nContagem por Gênero:")
    print(genero_counts)
```

```python
# Análise de Idade
if 'Idade' in df_2024.columns:
    plt.figure(figsize=(10, 6))
    df_2024['Idade'].hist(bins=20, edgecolor='black', alpha=0.7)
    plt.xlabel('Idade')
    plt.ylabel('Frequência')
    plt.title('Distribuição de Idade - 2024')
    plt.tight_layout()
    plt.show()
    
    print("\nEstatísticas de Idade:")
    print(df_2024['Idade'].describe())
```

```python
# Comparação entre anos - Quantidade de registros
years = ['2022', '2023', '2024']
counts = [len(df_2022), len(df_2023), len(df_2024)]

plt.figure(figsize=(10, 6))
plt.bar(years, counts, color=['#2ecc71', '#3498db', '#9b59b6'], edgecolor='black')
plt.xlabel('Ano')
plt.ylabel('Quantidade de Registros')
plt.title('Quantidade de Registros por Ano')
for i, v in enumerate(counts):
    plt.text(i, v + 50, str(v), ha='center', fontweight='bold')
plt.tight_layout()
plt.show()
```

```python
# Correlação entre variáveis numéricas - 2024
numeric_cols = df_2024.select_dtypes(include=[np.number]).columns
print(f"Colunas numéricas: {list(numeric_cols)}")

if len(numeric_cols) > 1:
    plt.figure(figsize=(14, 10))
    sns.heatmap(df_2024[numeric_cols].corr(), annot=True, cmap='coolwarm', center=0, fmt='.2f')
    plt.title('Mapa de Correlação - Variáveis Numéricas 2024')
    plt.tight_layout()
    plt.show()
```

```python
# Análise por Fase/Turma
if 'Fase' in df_2024.columns:
    fase_counts = df_2024['Fase'].value_counts()
    
    plt.figure(figsize=(10, 6))
    fase_counts.plot(kind='bar', color='coral', edgecolor='black')
    plt.xlabel('Fase')
    plt.ylabel('Quantidade')
    plt.title('Distribuição por Fase - 2024')
    plt.xticks(rotation=0)
    plt.tight_layout()
    plt.show()
```

```python
# Resumo final
print("=" * 60)
print("RESUMO DA ANÁLISE EXPLORATÓRIA")
print("=" * 60)
print(f"\nDataset PEDE2024:")
print(f"  - Total de registros: {len(df_2024)}")
print(f"  - Total de colunas: {len(df_2024.columns)}")
print(f"  - Valores nulos: {df_2024.isnull().sum().sum()}")
if 'INDE 2024' in df_2024.columns:
    print(f"  - INDE 2024 médio: {df_2024['INDE 2024'].mean():.2f}")
if 'Idade' in df_2024.columns:
    print(f"  - Idade média: {df_2024['Idade'].mean():.1f} anos")
```
